```python
import requests

## API URL
url = 'https://apis.data.go.kr/B552845/perDay/price?serviceKey={0}&pageNo=1&numOfRows=200&returnType=JSON&cond[exmn_ymd::GTE]={1}0101&cond[exmn_ymd::LTE]={1}1231&cond[se_cd::EQ]=02&cond[ctgry_cd::EQ]=200&cond[item_cd::EQ]=211&cond[grd_cd::EQ]=04&cond[sgg_cd::EQ]=1101&cond[mrkt_cd::EQ]=0110211&cond[vrty_cd::EQ]={2}'

api_key = 'be1cc5a17f9f4a7790f151e42c5dbb6baff8839ddca49583173942c9e7ff58c9'

variety_names = {
    '01': '봄',
    '02': '여름(고랭지)',
    '03': '가을',
    '06': '월동'
}

all_data = []

for year in range(2015, 2025):
    for variety in variety_names:

        request_url = url.format(api_key, year, variety)

        response = requests.get(request_url)

        if response.status_code == 200:
            print(f'API 호출 성공 | {year}년 / {variety_names[variety]}')

        data = response.json()
        items = data['response']['body']['items']['item']

        if year == 2015 and variety == '01':
            print('수집된 가격 데이터 예시:')
            print(items[:3])

        print(f'{year}년 / {variety_names[variety]} → {len(items)}건')

        all_data.extend(items)

print('전체 수집 데이터:', len(all_data), '건')
```
